```python
from sklearn.model_selection import train_test_split

df = df.drop_duplicates().reset_index(drop=True)

X = df.drop('Class', axis=1)
y = df['Class']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print('X_train:', X_train.shape, '| y_train:', y_train.shape)
print('X_test :', X_test.shape,  '| y_test :', y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

num_cols = ['Time', 'Amount'] + [f'V{i}' for i in range(1,29)]
cat_cols = [] 

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols),
    ],
    remainder='drop'
)
```

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

pipeline = ImbPipeline([
    ('prep', preprocessor),
    ('smote', SMOTE(random_state=42)),
    ('clf', RandomForestClassifier(random_state=42)),
])
    
```

```python
pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)
print('Pipeline selesai di-fit dan prediksi berhasil.')
```

```python
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE

def make_pipeline(clf, use_smote=False, use_class_weight=False):
    steps = [('prep', preprocessor)]
    if use_smote:
        steps.append(('smote', SMOTE(random_state=42)))
    if use_class_weight:
        clf = clf.set_params(class_weight='balanced')
    steps.append(('clf', clf))
    return ImbPipeline(steps)

models = {
    'RF_baseline (no handling)'  : make_pipeline(RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42)),
    'RF_class_weight_balanced'   : make_pipeline(RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42), use_class_weight=True),
    'RF_SMOTE'                   : make_pipeline(RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42), use_smote=True),
    'XGB_SMOTE'                  : make_pipeline(XGBClassifier(scale_pos_weight=10, eval_metric='aucpr', random_state=42), use_smote=True),
    'LGBM_SMOTE'                 : make_pipeline(LGBMClassifier(is_unbalance=True, random_state=42, verbose=-1), use_smote=True),
}
```

```python
from sklearn.metrics import (precision_score, recall_score, f1_score,
                             roc_auc_score, average_precision_score)

results = {}
for name, mdl in models.items():
    print(f'Training {name} ...')
    mdl.fit(X_train, y_train)
    y_pred_ = mdl.predict(X_test)
    y_proba_ = mdl.predict_proba(X_test)[:, 1]
    results[name] = {
        'Precision (fraud)': precision_score(y_test, y_pred_, pos_label=1, zero_division=0),
        'Recall (fraud)'   : recall_score(y_test, y_pred_, pos_label=1, zero_division=0),
        'F1 (fraud)'       : f1_score(y_test, y_pred_, pos_label=1, zero_division=0),
        'ROC-AUC'          : roc_auc_score(y_test, y_proba_),
        'PR-AUC'           : average_precision_score(y_test, y_proba_),
    }

print('Selesai training semua model.')
```

```python
results_df = pd.DataFrame(results).T.round(4).sort_values('F1 (fraud)', ascending=False)
print(results_df)
```

```python
from sklearn.metrics import (classification_report, confusion_matrix,
                             roc_auc_score, precision_recall_curve,
                             average_precision_score, f1_score)

y_pred  = pipeline.predict(X_test)
y_proba = pipeline.predict_proba(X_test)[:, 1]

print('Confusion Matrix:'); print(confusion_matrix(y_test, y_pred))
print('\nClassification Report:'); print(classification_report(y_test, y_pred, digits=4))
print('ROC-AUC:', roc_auc_score(y_test, y_proba))
print('PR-AUC :', average_precision_score(y_test, y_proba))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm, annot=True, fmt='d', cmap='Blues',
    xticklabels=['Normal', 'Fraud'],
    yticklabels=['Normal', 'Fraud']
)
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.title('Confusion Matrix - RF + SMOTE')
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```
